```python
# Language recognization using finite state automata anad finite state machine

# M = ( Q, E, delta, q0, F)

# Machine that starts and end with a
vocabulary = ['a', 'b']
transitions = {
    ('q0', 'a'): 'q1',
    ('q0', 'b'): 'dead',
    ('q1', 'a'): 'q1',
    ('q1', 'b'): 'q2',
    ('q2', 'a'): 'q1',
    ('q2', 'b'): 'q2',
}

initial_state = 'q0'
final_state = {'q1'}

def analyze_string(string):
    current_state = initial_state

    for char in string:
        if (current_state, char) in transitions:
            current_state = transitions[(current_state, char)]
            print(f"Moved to: {current_state}")
        else:
            print("Invalid transition ")
            return 
    
    if current_state in final_state:
        print("String is accepted")
    else:
        print("String is not accepted")


string = input("Enter the string")
analyze_string(string)
```

```python
# About the mealy machine

transitions = {
    ('S0', '0'): ('S0', '0'),
    ('S0', '1'): ('S1', '0'),

    ('S1', '0'): ('S2', '0'),
    ('S1', '1'): ('S1', '0'),

    ('S2', '0'): ('S0', '0'),
    ('S2', '1'): ('S3', '1'),

    ('S3', '0'): ('S2', '0'),
    ('S3', '1'): ('S1', '0'),
}

def mealy_machine(string):
    string = string.lower()
    current_state = 'S0'
    path = [current_state]
    output = []
    
    for char in string:
        if(char not in ['0','1']):
            print("Invalid string")
        transition = (current_state,char)

        if transition in transitions:
            (current_state, op) = transitions[transition]
            path.append(current_state)
            output.append(op)
            print(f"Character: {char} moved to state {current_state}: {op}")

string = input("Enter the string")
print(f"Finding the %3 for binary string {string}")
mealy_machine(string)
```

```python
transitions = {
    ('S0', '0'): ('S0', '0'),
    ('S0', '1'): ('S1', '0'),

    ('S1', '0'): ('S2', '0'),
    ('S1', '1'): ('S1', '0'),

    ('S2', '0'): ('S0', '0'),
    ('S2', '1'): ('S3', '1'),

    ('S3', '0'): ('S2', '0'),
    ('S3', '1'): ('S1', '0'),
}

def mealy_machine(string):
    current_state = 'S0'
    output = []

    for char in string:
        if (current_state, char) in transitions:
            next_state, op = transitions[(current_state, char)]
            print(f"Input: {char}  |  State: {current_state} -> {next_state}  |  Output: {op}")
            current_state = next_state
            output.append(op)
        else:
            print("Invalid character")
            return

    print("Final Output:", ''.join(output))

string = input("Enter the binary string: ")
mealy_machine(string)
```
