```python
import nltk
from nltk import ngrams
from nltk.tokenize import word_tokenize
from collections import Counter
import string

sentence  = """"
Alphabet Chief Executive Sundar Pichai said no company would be unscathed if the artificial intelligence boom collapses, as soaring valuations and heavy investment in the sector fuel concerns of a bubble.

Pichai said in an interview with the BBC published on Tuesday that the current wave of AI investment was an "extraordinary moment" but acknowledged "elements of irrationality" in the market, echoing warnings of "irrational exuberance" during the dotcom era.

"""
```

```python
tokens = [
    words for words in word_tokenize(sentence)
    if words not in string.punctuation
]

print(tokens)
```

```python
unigrams = list(ngrams(tokens, 1))
bigrams = list(ngrams(tokens, 2))
trigrams = list(ngrams(tokens, 3))

print(unigrams)
print(bigrams)
print(trigrams)
```

```python
unigram_counts = Counter(unigrams)
bigram_counts = Counter(bigrams)
trigram_counts = Counter(trigrams)

print(unigram_counts)
print(bigram_counts)
print(trigram_counts)

total_unigrams = sum(unigram_counts.values())
print(total_unigrams)
```

```python
unigram_prob = {gram: count/total_unigrams for gram, count in unigram_counts.items()}

bigram_prob = {gram: count/ unigram_counts[(gram[0],)] for gram, count in bigram_counts.items()}

trigram_prob ={gram: count/ bigram_counts[(gram[0], gram[1] )] for gram, count in trigram_counts.items()}
```

```python
print("Unigram")
for gram, prob in bigram_prob.items():
    print(f"{gram} :{prob:.5f}")
```

```python
for gram , prob in trigram_prob.items():
    print(f"{gram}  : {prob:.3f}")
```

```python
for gram , prob in trigram_prob.items():
    print(f"{gram}: {prob:.5f}")
```
