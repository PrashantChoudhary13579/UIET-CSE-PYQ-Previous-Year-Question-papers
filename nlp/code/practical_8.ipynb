```python
import nltk
# import spacy
from nltk import Nonterminal, PCFG
from nltk.parse import InsideChartParser

grammar = PCFG.fromstring("""
S -> NP VP [1.0]
NP -> Det N [0.6] | Det N PP [0.4]
VP -> V NP [0.5] | V NP PP [0.3] | V PP [0.2]
PP -> P NP [1.0]

Det -> 'a' [0.2] | 'the' [0.8]
N -> 'cat' [0.5] | 'dog' [0.5]
V -> 'chased' [0.7] | 'sat' [0.3]
P -> 'with' [0.6] | 'in' [0.4]
                        
""")

parser = InsideChartParser(grammar)

sentence = 'The dog chased the cat with a dog'
print(f"Sentence - {sentence}")
# token_sent = [token.text for token in sentence if not token.is_stop and token.is_alpha]
sent = sentence.lower().split()
print("After splitting - ",sent)
print("For Parsing the sentence:", ''.join(sent))

for tree in parser.parse(sent):
    print("Parse tree")
    print(tree)
    tree.pretty_print()
```

```python
import nltk
from nltk.parse import FeatureChartParser
from nltk.grammar import FeatureGrammar


feature_grammar = FeatureGrammar.fromstring("""

S[NUM=?n] -> NP[NUM=?n] VP[NUM=?n]

NP[NUM=?n] -> DET[NUM=?n] N[NUM=?n] | N[NUM=?n]
VP[NUM=?n] -> V[NUM=?n] NP[NUM=?x] | V[NUM=?n] 

DET[NUM=sg] -> 'this' | 'a'
DET[NUM=pl] -> 'these' | 'those'

N[NUM=sg] -> 'dog' |'cat'|'book'
N[NUM=pl] -> 'dogs' |'cats'|'books'

V[NUM=sg] -> 'runs' | 'barks' | 'reads'
V[NUM=pl] -> 'run' | 'bark' | 'read'



""")

feature_parser = FeatureChartParser(feature_grammar)

sentence = ["this dog runs", "these dogs runs"]

print("Testing the subject verb agreement ")
for sent in sentence:
    tokens = sent.split()
    # print(f"Sentence: {token}")
    print(f"Sentence: {sent}")
    trees = list(feature_parser.parse(tokens))
    # print(trees)
    if trees:
        print("Valid statement")
        for tree in trees:
            print(tree)
    else:
        print("Not a valid statement")
    print()
```
