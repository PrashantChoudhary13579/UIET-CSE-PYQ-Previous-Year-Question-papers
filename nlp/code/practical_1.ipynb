```python
import spacy
import nltk
from nltk.stem import PorterStemmer

nlp = spacy.load("en_core_web_sm")

text = "Hi siri, I am learning NLP."
doc = nlp(text)
stemmer = PorterStemmer()
# Tokenization
tokens = [token.text for token in doc]
print(tokens)
print(doc)

filter_token = [ token.text for token in doc if not token.is_stop and token.is_alpha]
print("Filtered tokens - ",filter_token)

print("Stemming- ")
for token in tokens:
    print(f"{token:20} -> {stemmer.stem(token)}")

print("Lemmatization: ")
for token in doc:
    print(f"{token.text:20} -> {token.lemma_}")

print("POS Tagging: ")
for token in doc:
    print(f"{token.text:20} -> {token.pos_:10} -> {token.tag_}")
```

```python
print("Noun phase chunking ")
for chunk in doc.noun_chunks:
    print(f"{chunk.text}")

print("Syntax Dependency parsing")
for token in doc:
    print(f"{token.text:20} -> {token.dep_:15} -> {token.head.text}")

print("NER")
for ent in doc.ents:
    print(f"{ent.text:15} -> {ent.label_}")
```
